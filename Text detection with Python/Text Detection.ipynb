# EasyOCR – Deep learning OCR for natural images and multilingual recognition.

```python
import easyocr
import cv2

# Read image
image = cv2.imread("Image 01.png")

# Initialize reader
reader = easyocr.Reader(['en'])

# Detect text
results = reader.readtext(image)

# Print text
for result in results:
    print("Detected:", result[1])
```

# Draw Bounding Boxes

```python
import cv2
import easyocr
import matplotlib.pyplot as plt

# Read image
image = cv2.imread("Image 01.png")

# OCR Reader
reader = easyocr.Reader(['en'])

# Detect text
results = reader.readtext(image)

# Draw bounding boxes
for (bbox, text, prob) in results:

    top_left = tuple(map(int, bbox[0]))
    bottom_right = tuple(map(int, bbox[2]))

    # Rectangle
    cv2.rectangle(image,
                  top_left,
                  bottom_right,
                  (0,255,0),
                  2)

    # Text
    cv2.putText(image,
                text,
                top_left,
                cv2.FONT_HERSHEY_SIMPLEX,
                0.8,
                (0,0,255),
                2)

# Convert BGR to RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Show image
plt.figure(figsize=(12,8))
plt.imshow(image_rgb)
plt.axis("off")
plt.show()
```
